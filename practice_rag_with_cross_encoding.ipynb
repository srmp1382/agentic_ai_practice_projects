```python
from pypdf import PdfReader
```

```python
reader=PdfReader('/content/sample_data/ncert.pdf')
```

```python
total_pages=len(reader.pages)
```

```python
total_pages
```

```python
text=[]
for i in range(1,15):
  t=reader.pages[i]
  text.append(t.extract_text())
```

```python
print(text)
```

```python
len(text)
```

```python
parsed_text=' '.join(text)
```

```python
parsed_text
```

```python
import re
```

```python
clean_text=re.sub(r'(?<=\w)\s*\r?\n\s*(?=\w)', ' ', parsed_text) #this eliminates the \n and \r inside sentences and preserves the paragraph breaks only.
```

```python
clean_text
```

```python
pip install langchain langchain-core langchain-text-splitters
```

```python
import langchain_text_splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter=RecursiveCharacterTextSplitter(separators=['.','\r','\n'])
```

```python
chunks=splitter.split_text(clean_text)
```

```python
len(chunks)
```

```python
pip install chromadb
```

```python
import chromadb
```

```python
client=chromadb.PersistentClient(path='.\\my_chromadb')
```

```python
pip install sentence-transformers
```

```python
import sentence_transformers
```

```python
from sentence_transformers import SentenceTransformer
```

```python
embed_model=SentenceTransformer('all-MiniLM-L6-V2')
```

```python
embeddings=embed_model.encode(chunks)
```

```python
collection=client.get_or_create_collection(name='my_collection', embedding_function=None)
```

```python
ids=list(map(lambda x: 'chunk-'+str(x),list(range(len(chunks)))))
```

```python
ids
```

```python
collection.add(ids=ids, documents=chunks, embeddings=embeddings)
```

```python
query='What is the difference between web of things and grid computing?'
```

```python
query_embedding=embed_model.encode(query)
```

```python
result=collection.query(query_embeddings=[query_embedding], n_results=10, include=['documents','distances'])
```

```python
len(result)
```

```python
print(result['distances'])
```

```python
print(result['documents'][0][0])
```

```python
from sentence_transformers import CrossEncoder
```

```python
cross_encoder=CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-V2')
```

```python
pairs=[[query, doc] for doc in result['documents'][0]]
```

```python
query
```

```python
scores=cross_encoder.rank(query, result['documents'][0])
```

```python
for i in zip(result['ids'][0],result['distances'][0]):
  print(list(i))
```

```python
ids_after_cross=[]
for id, score_ce in enumerate(scores):
  ids_after_cross.append(score_ce['corpus_id'])
```
